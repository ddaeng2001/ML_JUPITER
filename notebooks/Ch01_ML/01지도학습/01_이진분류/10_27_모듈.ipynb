# 라이브러리 가져오기

```python
import pandas as pd
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
```

# EDA

```python
train.head() # 종속변수 income 포함되어있음
test.head()

train.shape #(29304, 16)
test.shape #(3257, 15)

train.isnull().sum() 
# age(12), workclass(1662), occupationg(1668), hours.per.week(13), native.country(537)

test.isnull().sum()
# age(6), workclass(174), occupation(175), hours.per.week(9), native.country(46)
```

# 결측치 채우기(train.ver)

```python
#-------------------------
# workclass와 occupation
# -------------------------
# workclasss와 occupation 결측치가 많고 범주형이기 때문에
# 단순 최빈값 대체는 데이터 편향 발생 O & 결측치 자체가 중요한 정보(ex> 정보없음)이거나 특정 의미를 내포하고 있을수도 있음
# 따라서 새로운 카테고리('UNKNOWN')으로 분류
train['workclass'] = train['workclass'].fillna('UNKNOWN')
train['occupation'] = train['occupation'].fillna('UNKNOWN')

# workclass와 occupation의 결측치 처리 완료 #결과 : 0
train['workclass'].isnull().sum()
train['occupation'].isnull().sum()




#-------------------------
# native.country
#-------------------------
# native.country컬럼 내 United-States 값들인 경우의 개수
count_united_states = train['native.country'].value_counts()['United-States']
count_united_states # 29304개 중 26240개

# 범주형이며 United-States의 값이 압도적이므로 결측치를 최빈값으로 대체해도
# 데이터 자체가 이미 극도로 편향되어있기 때문에 영향을 끼치지 않음

print(f"native.country 최빈값: {train['native.country'].mode()[0]}") # native.country 컬럼의 최빈값 확인
train['native.country'] = train['native.country'].fillna('United-States') # 최빈값으로 결측치 채우기
train['native.country'].isnull().sum() # 결측치 처리 완료 - 결과 : 0




#-------------------------
# age와 hours.per.week
#-------------------------
# 두 개의 컬럼은 결측치가 매우 낮고 수치형이지만 소득예측에서 중요한 독립변수이기에 중앙값으로 대체함

value = int(train['age'].median()) # 중앙값 구하기
print("중앙값 : " ,value) #결과 : 37
train['age'] = train['age'].fillna(value) # 결측치를 중앙값으로 채우기
train['age'].isnull().sum() # 결측치 처리 완료 - 결과 : 0


value = int(train['hours.per.week'].median()) # 중앙값 구하기
print("중앙값 : " ,value) #결과 : 40
train['hours.per.week'] = train['hours.per.week'].fillna(value) # 결측치를 중앙값으로 채우기
train['hours.per.week'].isnull().sum() # 결측치 처리 완료 - 결과 : 0


```
