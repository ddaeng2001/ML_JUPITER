```python
# Matplotlib 단계별 예시 및 활용
# Matplotlib은 Python에서 정적(Static)이고 고품질의 그래프를 생성하는 데 가장 널리 사용되는 기본 라이브러리입니다.
# 모든 다른 Python 시각화 라이브러리(Seaborn, Pandas plot 등)의 기반이 됩니다.

# Matplotlib을 써야 하는 경우
# 연구 및 보고서용 그래프: 논문, 보고서, 프레젠테이션 등에 사용할 정확하고 세밀하게 제어된 이미지 파일 형태의 그래프가 필요할 때.
# 기초적인 탐색적 분석 (EDA): 데이터의 분포나 두 변수 간의 관계를 빠르게 확인할 때 (히스토그램, 산점도 등).
# 세밀한 커스터마이징: 축 레이블, 제목, 색상, 폰트 등 그래프의 모든 요소를 가장 자유롭게 제어해야 할 때.
# 다른 라이브러리의 기반 작업: 다른 고수준 라이브러리(예: Seaborn)로 그래프를 그린 후, 추가적인 미세 조정이 필요할 때.
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


# 데이터 준비 (시간에 따른 주가 변화 가정)
x = np.linspace(0, 10, 100) # 0부터 10까지 100개의 점
y = np.sin(x) + np.random.normal(0, 0.1, 100)

df = pd.read_csv("CC GENERAL.csv")
x = df['CUST_ID']
y = df['PAYMENTS']

# 그래프 그리기
plt.figure(figsize=(8, 4)) # 캔버스(Figure) 크기 설정
plt.plot(x, y, label='label', color='blue', linestyle='--') # 선 그래프 그리기
plt.title('title') # 제목
plt.xlabel('x_axis') # X축 레이블
plt.ylabel('y_axis') # Y축 레이블
plt.legend('bum') # 범례 표시
plt.grid(True) # 그리드 표시
plt.show() # 그래프 출력
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터 준비 (사람들의 나이 분포 가정)
data_age = np.random.normal(loc=35, scale=10, size=500) # 평균 35, 표준편차 10
type(data_age)


# # plt.figure(figsize=(8, 4))
# # # 히스토그램 그리기
plt.hist(data_age, bins=20, edgecolor='black', color='red', alpha=0.5) 
plt.title('title')
plt.xlabel('x')
plt.ylabel('y')
plt.axvline(np.mean(data_age), color='red', linestyle='dashed', linewidth=1, label=f'mean: {np.mean(data_age):.2f}') # 평균선 추가
plt.legend()
plt.show()
```

```python
# 데이터 준비 (광고비 지출과 매출액 가정)
ad_spend = np.random.uniform(100, 1000, 50)
sales = ad_spend * 2.5 + np.random.normal(0, 500, 50) # 양의 상관관계

plt.figure(figsize=(8, 4))
# 산점도 그리기
plt.scatter(ad_spend, sales, color='purple', alpha=0.6, s=50) # s는 점의 크기
plt.title('title')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
